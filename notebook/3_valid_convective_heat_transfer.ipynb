```python
import sys
sys.path.append('../')
from matplotlib import pyplot as plt
import numpy as np

from ventilation_layer import heat_transfer_coefficient as htc
```

# 対流熱伝達率の検証（有風）

```python
# 風速をパラメータにする
v_a = [x / 200.0 for x in range(100)]
```

```python
# 温度条件は固定
theta_1 = 20.0
theta_2 = 16.0
```

```python
# 角度はパラメータ
angle = [x * 1.0 for x in range(91)]
```

```python
# 通気層高さ、厚さは固定
l_h = 3.0
l_d = 0.03
```

```python
# 角度の影響を見る

h_cv01_AK = np.vectorize(htc.get_h_cv)(calc_mode="detailed", v_a=0.1, theta_1=theta_1, theta_2=theta_2, angle=angle, l_h=l_h, l_d=l_d)
h_cv05_AK = np.vectorize(htc.get_h_cv)(calc_mode="detailed", v_a=0.5, theta_1=theta_1, theta_2=theta_2, angle=angle, l_h=l_h, l_d=l_d)
h_cv10_AK = np.vectorize(htc.get_h_cv)(calc_mode="detailed", v_a=1.0, theta_1=theta_1, theta_2=theta_2, angle=angle, l_h=l_h, l_d=l_d)

h_cv01_JU = np.full(91, htc.get_h_cv_Jurges(v_a=0.1))
h_cv05_JU = np.full(91, htc.get_h_cv_Jurges(v_a=0.5))
h_cv10_JU = np.full(91, htc.get_h_cv_Jurges(v_a=1.0))
```

```python
# グラフの描画
fig1 = plt.figure()
ax1 = fig1.subplots()

ax1.set_xlabel('angle[-]')
ax1.set_ylabel('convective heat transfer[W/(m2 K)]')

plt.grid(which='major',color='gray',linestyle='-')
plt.grid(which='minor',color='gray',linestyle='-')

plt.plot(angle, h_cv01_AK, linestyle='solid', color='r', label='This Method(Akasaka), v_a=0.1m/s')
plt.plot(angle, h_cv05_AK, linestyle='solid', color='b', label='This Method(Akasaka), v_a=0.5m/s')
plt.plot(angle, h_cv10_AK, linestyle='solid', color='g', label='This Method(Akasaka), v_a=1.0m/s')
plt.plot(angle, h_cv01_JU, linestyle='dashed', color='r', label='Jurges, v_a=0.1m/s')
plt.plot(angle, h_cv05_JU, linestyle='dashed', color='b', label='Jurges, v_a=0.5m/s')
plt.plot(angle, h_cv10_JU, linestyle='dashed', color='g', label='Jurges, v_a=1.0m/s')

plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.show()
```
